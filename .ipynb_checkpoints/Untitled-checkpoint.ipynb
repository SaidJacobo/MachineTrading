```python
import pandas as pd
import re
pd.set_option('display.max_rows', None)
```

```python
df = pd.read_csv('./results.csv')
df
```

```python
a = 'mode_train-model_gradient_boosting-trainwindow_114-trainperiod_1-tradingstrategy_strategies.bband_strategy'

re.search(r'tradingstrategy_strategies\.(\w+_strategy)', a).group(1)
```

```python
df['model'] = df['experiment'].apply(lambda x: re.search(r'-model_(.*?)-', x).group(1))
df['window'] = df['experiment'].apply(lambda x: re.search(r'-trainwindow_(.*?)-', x).group(1))
df['train_period'] = df['experiment'].apply(lambda x: re.search(r'-trainperiod_(.*?)-', x).group(1))

df['strategy'] = df['experiment'].apply(lambda x: re.search(r'tradingstrategy_strategies\.(\w+_strategy)', x).group(1))
```

```python
df
```

```python
gb = df.groupby(
    by=['model','strategy']
).agg(
    {
        'operation_ratio_2':['mean', 'std']
    }
).reset_index()


# codigo para que se vea lindo
gb.columns = gb.columns.droplevel(0)
gb.columns = ['model','strategy','mean','std']
gb['model_strategy'] = df['model'] + '-' + df['strategy']
gb
```

```python
gb = df.groupby(
    by=['model','strategy', 'window', 'train_period']
).agg(
    {
        'operation_ratio_2':['mean', 'std']
    }
)

gb
```

```python
df.groupby(by=['model','strategy']).agg(
    {
        'operation_ratio_2':['mean', 'std']
    }
).reset_index().sort_values(by=('operation_ratio_2','mean'), ascending=False)
```

```python
df[
    (df['model'].isin(['random_forest','gradient_boosting','neural_network']))
    & (df['strategy'].isin(['bband_strategy']))

].groupby(by=['model','strategy', 'window']).agg({'operation_ratio_2':['mean', 'std']})
```
